```python
from qiskit import QuantumCircuit
from qiskit import transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import (NoiseModel,
                              depolarizing_error,
                              pauli_error)
import matplotlib.pyplot as plt
```

```python
backend = AerSimulator()

noise_model = NoiseModel.from_backend(backend)

error = depolarizing_error(0.05, 1)
noise_model.add_all_qubit_quantum_error(error, ["x", "z", "h"])
error_gate1 = pauli_error([("X", 0.05), ("I", 1 - 0.05)])
error_gate2 = error_gate1.tensor(error_gate1)
noise_model.add_all_qubit_quantum_error(error_gate2, ["cx"])

sim_noisy = AerSimulator(noise_model = noise_model)

q = QuantumCircuit()
```

```python
#X n2
```

```python
path = "./qasm_files/polar_n2_x.qasm"
qc = q.from_qasm_file(path)
qc.draw('mpl', fold = -1)
```

```python
ncirc = transpile(qc)
result = backend.run(ncirc).result()
counts = result.get_counts()

# print(counts)

x = []
y = []

l = len(qc.clbits)
l = pow(10, l-4)
for i in counts:
    a = int(i)
    if (a % l == 0):
        print(counts[i], i[0:4])
        x.append(str(i[0:4]))
        y.append(counts[i])
```

```python
plt.bar(x, y)
plt.show()
```

```python
ncirc = transpile(qc, sim_noisy)
result = sim_noisy.run(ncirc).result()
counts = result.get_counts()

# print(counts)

x = []
y = []

l = len(qc.clbits)
l = pow(10, l-4)
for i in counts:
    a = int(i)
    if (a % l == 0):
        print(counts[i], i[0:4])
        x.append(str(i[0:4]))
        y.append(counts[i])
```

```python
plt.bar(x, y)
plt.show()
```
